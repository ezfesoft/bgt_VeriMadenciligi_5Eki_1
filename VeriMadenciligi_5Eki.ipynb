```python
import pandas as pd
df = pd.read_csv('ogrenci_veriseti_500.csv')
print("--- İlk 5 Satır ---")
print(df.head())
print("\n--- Sütun Başına Eksik Veri Sayısı ---")
print(df.isna().sum())
print("\n--- İstatistiksel Özet ---")
print(df.describe())
```

```python
df_temiz=df.dropna()
print(df_temiz.describe())
```

```python
df_temiz = df.dropna()

print("Silme işleminden sonra veri boyutu:", df_temiz.shape)
print(df_temiz)
```

```python
df_dolu = df.copy()

vize_ortalamasi = df_dolu['Vize_Notu'].mean()
df_dolu['Vize_Notu'] = df_dolu['Vize_Notu'].fillna(vize_ortalamasi)

yas_medyani = df_dolu['Yas'].median()
df_dolu['Yas'] = df_dolu['Yas'].fillna(yas_medyani)

df_dolu = df_dolu.fillna(0)

print("--- Doldurma İşleminden Sonra Veri Seti ---")
print(df_dolu)
print("\nKalan Eksik Veri Sayısı:\n", df_dolu.isna().sum())
```
